# Правило: когда использовать Restart / Run all

**Правило:** Restart полностью очищает runtime, а Run all выполняет ячейки сверху вниз в текущем состоянии. Restart и Run all нужен перед сдачей, после перестановки ячеек, изменения кода, переменных и для проверки воспроизводимости. Run all без Restart нужен для быстрого вывода, когда состояние уже проверено.

# Повышенный уровень: сравните два способа восстановления после ошибки порядка ячеек и объясните, какой безопаснее для новичка.

Ручной способ: надо выполнить ячейку с переменной, затем print. Этот способ быстрый, но ненадежный. В Runtime могут остаться старые переменные или можно пропустить важные аспекты, если код уже достаточно длинный.

Restart и Run all: очищает память и выполняет всё заново. Этот способ дольше, но надёжнее. Несмотря на время, полностью очищает Runtime и показывает, работает ли notebook.
Новичку безопаснее второй. Он не оставляет скрытых переменных, сразу показывает, где нарушен порядок и позволяет оценить производительность.

# Повышенный уровень

**Notebook survival guide**
| № | Проблема | Симптом | Диагностика | Решение |
|---|----------|---------|-------------|---------|
| 1 | **NameError** | name '...' is not defined | Проверить, выполнена ли ячейка, создающая переменную | Выполнить ячейку с присваиванием до её использования; Restart & Run all |
| 2 | **Опечатка в имени переменной** | NameError для переменной, которая есть | Сравнить имя в присваивании и в использовании | Исправить написание: project_title и projet_title |
| 3 | **Лишние скобки** | SyntaxError: unmatched ')' | Найти в строке ошибки лишнюю или пропущенную скобку | Сверить количество скобок, убрать лишнюю |
| 4 | **Нарушен порядок ячеек** | Часть кода работает, а другая нет | Запустить всё заново сверху вниз | Переставить ячейки в логическом порядке; Restart & Run all |
| 5 | **Имя переменной начинается с цифры** | SyntaxError: invalid synta` | Посмотреть на имя переменной | Начинать имя с буквы: 1meow на meow |
| 6 | **Незакрытая кавычка** | SyntaxError: unterminated string literal | Проверить, есть ли закрывающая кавычка в строке | Добавить закрывающую кавычку: вместо "текст - "текст" |
| 7 | **= вместо ==** | SyntaxError | Посмотреть на строку с if, while | Заменить = на == : if x = 5: на if x == 5: (это разное!!!) |
